```python
df=pd.read_csv("D:\\Project DS\\traffic_stops - traffic_stops_with_vehicle_number.csv")
df.head()
```

```python
print(df['search_conducted'].unique())
```

```python
print(df.dtypes)
```

```python
df['stop_date'] = pd.to_datetime(df['stop_date'],format='%Y-%m-%d',errors='coerce') 
```

```python
print(df.dtypes)
```

```python
df['stop_time'] = pd.to_datetime(df['stop_time'],format='%H:%M:%S',errors='coerce').dt.time
```

```python
print(df.dtypes)
```

```python
print(df['stop_time'].unique())
```

```python
df = df.dropna(axis=1, how='all')
```

```python
for col in['driver_age_raw', 'driver_age']:
    if col in df.columns:
        df[col]=df[col].fillna(df[col].median())
```

```python
df.head()
```

```python
for col in['search_conducted','is_arrested','drugs_related_stop']:
    if col in df.columns:
        df[col]=df[col].fillna(False).astype(bool)
```

```python
for col in df.select_dtypes(include='object').columns:
    if df[col].isnull().sum() > 0:
        df[col]=df[col].fillna(df[col].mode()[0])                    
```

```python
print(df.isnull().sum())
```

```python
print(df.dtypes)
```

```python
print(df.head())
```

```python
df.to_csv("clean_traffic_stops.csv",index="False")
```

```python
import pandas as pd
```

```python
df=pd.read_csv("clean_traffic_stops.csv")
```

```python
df.head()
```

```python
#INSER DATA INTO MYSQL
for index, row in df.iterrows():
    sql = """
    INSERT INTO traffic_stops (
        stop_date, stop_time, country_name, driver_gender, driver_age_raw,
        driver_age, driver_race, violation_raw, violation, search_conducted,
        search_type, stop_outcome, is_arrested, stop_duration, drugs_related_stop,
        vehicle_number
    ) VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
    """
    values = (
        row['stop_date'], row['stop_time'], row['country_name'], row['driver_gender'],
        row['driver_age_raw'], row['driver_age'], row['driver_race'], row['violation_raw'],
        row['violation'], row['search_conducted'], row['search_type'], row['stop_outcome'],
        row['is_arrested'], row['stop_duration'], row['drugs_related_stop'], row['vehicle_number']
    )
    cursor.execute(sql, values)

conn.commit()
cursor.close()
conn.close()
print("✅ Data inserted into MySQL.")
```
